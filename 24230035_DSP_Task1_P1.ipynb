```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

all_data = []  # Stores the results from all pages

for url in urls:
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        # Extracting Year and Quarter from URL
        match = re.search(r'(\d{4})Q(\d)', url)
        year = match.group(1) if match else None
        quarter = match.group(2) if match else None

        # for looping through each sale record
        for table in soup.find_all("table", class_="sale"):
            sale_info = {"Year": year, "Quarter": quarter}  # Store extracted values

            # Extracting the  details from each row
            rows = table.find_all("tr")
            for row in rows:
                cells = row.find_all("td")
                if len(cells) == 2:
                    key = cells[0].get_text(strip=True).replace(":", "")
                    value = cells[1].get_text(strip=True)
                    sale_info[key] = value

            # Checking "Customer Details" exists before processing
            customer_details = sale_info.get("Customer Details", "")

            if customer_details:
                # Extracting the customer details using regex
                id_match = re.search(r"ID:\s*(\d+)", customer_details)
                location_match = re.search(r"Location:\s*([\w\s]+?)(?=\s*Gender:|\s*Age Category:|$)", customer_details)
                gender_match = re.search(r"Gender:\s*(Male|Female)", customer_details)
                age_match = re.search(r"Age Category:\s*([\d\-]+)", customer_details)

                # Store extracted details
                sale_info["Customer ID"] = id_match.group(1) if id_match else None
                sale_info["Location"] = location_match.group(1).strip() if location_match else None
                sale_info["Gender"] = gender_match.group(1) if gender_match else None
                sale_info["Age Category"] = age_match.group(1) if age_match else None

                # Remove original "Customer Details" column
                sale_info.pop("Customer Details", None)

            # Append valid data
            if sale_info:
                all_data.append(sale_info)

    else:
        print(f"Failed to fetch {url}: {response.status_code}")

```

```python
# Convert to DataFrame and saveing in a csv file
df = pd.DataFrame(all_data)
df.to_csv("scraped_data.csv", index=False)
print(" Data saved to scraped_data.csv")
```
